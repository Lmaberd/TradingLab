```python
import pandas as pd, httpx, zipfile, io

resp = httpx.get("https://www.cftc.gov/files/dea/history/com_fin_txt_2026.zip", follow_redirects=True, timeout=30)

# zf checks for the file names available in the zip, which is FinComYY.txt
zf = zipfile.ZipFile(io.BytesIO(resp.content))
print(zf.namelist())

# Opens first file in the zip, low_memory arg tells pandas to be more cautious about column types
df = pd.read_csv(zf.open(zf.namelist()[0]), low_memory=False)
df
```

```python
names = df["Market_and_Exchange_Names"].unique()
[n for n in names if "S&P" in n.upper()]
```

```python
cols = ["Lev_Money_Positions_Long_All", "Lev_Money_Positions_Short_All",
        "Asset_Mgr_Positions_Long_All", "Asset_Mgr_Positions_Short_All",
        "Dealer_Positions_Long_All", "Dealer_Positions_Short_All"]

es = df[df["Market_and_Exchange_Names"].str.contains("E-MINI S&P 500", case=False)]
es[cols].isna().sum()
```
